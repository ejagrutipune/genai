## One-hot encoding â€” a simple text example

**Aim: Convert words into numbers.**

Each word gets its own column. **1 means this is the word; 0 means it is not.**

Run each cell from top to bottom using **Shift + Enter**.

```python
# Run only if scikit-learn is not installed:
# %pip install scikit-learn
```

### 1. Import the encoder

```python
from sklearn.preprocessing import OneHotEncoder
```

### 2. Take one sentence

```python
document = ["I love AI and I love Python"]
document
```

### 3. Split the sentence into words

`lower()` makes the text lowercase. `split()` separates words at spaces.

```python
tokens = [sentence.lower().split() for sentence in document]
tokens
```

Expected words: `['i', 'love', 'ai', 'and', 'i', 'love', 'python']`

### 4. Put each word in its own row

The encoder expects a table. Here, each row contains one word.

```python
all_words = [[word] for sentence in tokens for word in sentence]
all_words
```

### 5. Create the encoder

`sparse_output=False` gives us a regular array that is easy to display. `dtype=int` shows 0 and 1.

```python
encoder = OneHotEncoder(sparse_output=False, dtype=int)
```

### 6. Learn the vocabulary

`fit()` learns the unique words. Repeated words do not create extra columns.

```python
encoder.fit(all_words)
print("Vocabulary:", encoder.categories_[0])
```

The column order is: **ai, and, i, love, python**.

### 7. Convert the words into numbers

`transform()` converts each word using the learned vocabulary.

```python
encoded_words = encoder.transform(all_words)
print(encoded_words)
```

### 8. Show each word beside its vector

```python
print("Columns:", encoder.categories_[0])
print()

for word, vector in zip(all_words, encoded_words):
    print(word[0], "->", vector)
```

### Read one row

| Word | ai | and | i | love | python |
|---|---:|---:|---:|---:|---:|
| love | 0 | 0 | 0 | 1 | 0 |

Only the **love** column is 1. That is why it is called **one-hot**.

Both occurrences of **love** get the same vector. Each row represents **one word**, not the whole sentence.

### 9. Try a word yourself

Change `word` to ai, and, i, love, or python and rerun.

```python
word = "python"
print("Columns:", encoder.categories_[0])
print(word, "->", encoder.transform([[word]])[0])
```

### Remember

- **fit:** learn the vocabulary.
- **transform:** convert words using that vocabulary.
- **One-hot:** one position is 1; the others are 0.

This encoder uses its default behavior: a new word such as **cricket** will raise an error because it was not in the learned vocabulary.

**Practice:** Change the original sentence and rerun from step 2. Watch how the vocabulary and vectors change.

## Example 2: Encode city names

The same idea works for cities. Each row below represents a house.

We will replace **City** with yes/no columns and keep **Area** unchanged.

```python
# Run only if pandas is missing:
# %pip install pandas

import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

### 1. Create our house table

```python
houses = pd.DataFrame({
    "City": ["Pune", "Mumbai", "Delhi", "Pune"],
    "Area": [800, 1000, 900, 1200]
})
houses
```

### 2. Select the City column

Double brackets keep the input as a table, which the encoder expects.

```python
cities = houses[["City"]]
cities
```

### 3. Learn the city names

Use a separate encoder so the word example above stays unchanged.

```python
city_encoder = OneHotEncoder(sparse_output=False, dtype=int)
city_encoder.fit(cities)

print("Cities:", city_encoder.categories_[0])
```

### 4. Convert the cities into numbers

The columns are ordered **Delhi, Mumbai, Pune**.

```python
city_vectors = city_encoder.transform(cities)
print(city_vectors)
```

### 5. Add readable column names

```python
city_columns = city_encoder.get_feature_names_out(["City"])

city_table = pd.DataFrame(
    city_vectors,
    columns=city_columns,
    index=houses.index
)
city_table
```

### Read the table

| Original city | City_Delhi | City_Mumbai | City_Pune |
|---|---:|---:|---:|
| Pune | 0 | 0 | 1 |
| Mumbai | 0 | 1 | 0 |
| Delhi | 1 | 0 | 0 |
| Pune | 0 | 0 | 1 |

**1 = yes, 0 = no.** Both Pune houses get the same city vector.

### 6. Keep Area and replace City

Area is already numeric. It does not need one-hot encoding.

```python
encoded_houses = pd.concat([houses[["Area"]], city_table], axis=1)
encoded_houses
```

### 7. Try one city

Change the city to Pune, Mumbai, or Delhi and rerun.

```python
new_house = pd.DataFrame({"City": ["Mumbai"]})

print("Columns:", city_columns)
print(city_encoder.transform(new_house))
```

For Mumbai, the result is **[0, 1, 0]**.

A city not learned during fitting, such as Chennai, will raise an error with this encoder's default settings.

**Practice:** Change the first house's city from Pune to Delhi in step 1. Rerun the city example and check how the first row changes.

**Same idea in both examples:** one column for each known category, with 1 in the matching position.
