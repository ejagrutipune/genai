# Bag of Words — a simple example

**Bag of Words (BoW)** converts sentences into numbers by counting words. Most traditional machine-learning models need numbers as input.

We use the ready-made **CountVectorizer** library from scikit-learn. No pandas or manual counting code is needed.

Run the cells from top to bottom using **Shift+Enter** or **Run All**. Select a Python kernel first. If scikit-learn is missing, run `%pip install scikit-learn` in a new code cell.

## Step 1: Import the library

`CountVectorizer` builds the vocabulary and counts the words for us.

```python
from sklearn.feature_extraction.text import CountVectorizer
```

## Step 2: Create a few sentences

Each sentence is called a **document**. These movie/cricket examples follow your `encoding.ipynb`.

```python
documents = [
    'people watch movie',
    'people watch cricket',
    'people watch movie movie'
]

for sentence in documents:
    print(sentence)
```

## Step 3: Convert text into word counts

`fit_transform()` learns the unique words (the **vocabulary**) and counts them in each sentence.

```python
bow = CountVectorizer()
bow_matrix = bow.fit_transform(documents)
```

## Step 4: Display the vocabulary and numbers

Each **column** represents a vocabulary word. Each **row** represents a sentence, in the same order as `documents`.

`toarray()` lets us see the numbers in this small example.

```python
print('Vocabulary:')
print(bow.get_feature_names_out())

print('\nBag of Words matrix:')
print(bow_matrix.toarray())
```

## Step 5: Understand the result

The column order is **[cricket, movie, people, watch]**.

| Sentence | cricket | movie | people | watch |
|---|---:|---:|---:|---:|
| people watch movie | 0 | 1 | 1 | 1 |
| people watch cricket | 1 | 0 | 1 | 1 |
| people watch movie movie | 0 | 2 | 1 | 1 |

For the last sentence, **[0, 2, 1, 1]** means: zero cricket, two movie, one people, and one watch.

## Step 6: Convert a new sentence

Use `transform()` to keep the **same vocabulary and column order**. Do not fit again for new input.

```python
new_documents = ['people watch cricket cricket']

print('Vocabulary:', bow.get_feature_names_out())
print('New sentence:', new_documents[0])
print('Word counts:', bow.transform(new_documents).toarray())
```

Expected counts: **[2, 0, 1, 1]**.

An unfamiliar word is ignored because it has no column in the learned vocabulary. Below, `football` is ignored, but `people` and `watch` are counted.

```python
print(bow.transform(['people watch football']).toarray())
```

## Step 7: How is it related to one-hot encoding?

Using the same vocabulary **[cricket, movie, people, watch]**:

- One-hot for the single word `movie`: **[0, 1, 0, 0]**.
- BoW for the document `movie movie`: **[0, 2, 0, 0]**.

One-hot identifies one word. BoW counts all word occurrences in a document. Adding the one-hot vectors for those occurrences produces its count-based BoW vector.

```python
print('Counts for movie movie:')
print(bow.transform(['movie movie']).toarray())
```

## Step 8: Remember these points

- **Where?** Spam detection, review classification, and article topics.
- **When?** When the words present in text provide useful clues.
- **Why a bag?** Basic BoW ignores word order: `people watch movie` and `movie watch people` have the same counts.
- **Limitation:** It does not understand meaning or recognize synonyms by itself.
- **Next:** Give these numbers to a classifier to learn labels. BoW itself does not make predictions.
